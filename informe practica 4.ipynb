<h5><p><strong>Componentes de un equipo de adquisición<strong><p><h5>

<p>Un sistema de adquisición de datos puede entenderse como una arquitectura modular compuesta por diferentes subsistemas que trabajan de manera integrada para captar, procesar y presentar información proveniente de un fenómeno físico. Estos sistemas se estructuran en cuatro módulos principales: 
<p>(1) el módulo sensor, encargado de interactuar con el sistema bajo estudio y convertir magnitudes físicas en señales medibles (por ejemplo, voltaje); (2) el módulo de acondicionamiento de señal, que incluye funciones analógicas como amplificación y filtrado, así como funciones digitales como temporización y digitalización; (3) el módulo de salida, que presenta la información en un formato interpretable; y (4) el módulo de control, que permite la retroalimentación y actuación sobre el sistema en configuraciones activas. [1]
Esta estructura general puede observarse en aplicaciones específicas como los sistemas de adquisición de señales electromiográficas (EMG), donde se implementan etapas equivalentes que incluyen electrodos como sensores, preprocesamiento (preamplificación y filtrado), procesamiento adicional de la señal, conversión analógico-digital, suministro de energía y, en muchos casos, módulos de transmisión inalámbrica. De este modo, los sistemas biomédicos no son más que una adaptación especializada de la arquitectura general de adquisición de datos aplicada a señales fisiológicas.<p> [2]

<p>Se consideran estáticas la interferencia de la red eléctrica (50/60 Hz), el ruido del sistema de adquisición y el desequilibrio de impedancias; sin embargo, no son completamente constantes, ya que pueden experimentar pequeñas variaciones en amplitud o condiciones de operación. En contraste, las fuentes variantes en el tiempo, como la interferencia electromagnética, la interfaz electrodo-piel, los artefactos de movimiento y la actividad muscular no deseada, dependen de factores dinámicos del entorno y del sujeto, presentando cambios impredecibles en amplitud y frecuencia.<p>

<h5><strong>3)<strong><h5>

<p>En el registro de biopotenciales de superficie, la capa más externa de la piel, el estrato córneo, desempeña un papel fundamental, ya que constituye la principal contribución a la impedancia total de la piel y domina su comportamiento en el rango de bajas frecuencias (1 Hz a 10 kHz) [4]. Esta característica convierte a la interfaz piel-electrodo en un elemento determinante en la calidad de la señal adquirida.<p>

<p>En este contexto, lograr una interfaz de baja impedancia entre el electrodo y la piel es esencial para garantizar una adecuada transducción de las señales fisiológicas. Una impedancia elevada incrementa el ruido térmico en los electrodos, lo que deteriora la relación señal-ruido (SNR) y afecta directamente la fidelidad de la señal registrada. Como consecuencia, pueden presentarse dificultades en la identificación de eventos fisiológicos relevantes, como la activación muscular en estudios de electromiografía (EMG), o alteraciones morfológicas en señales electrocardiográficas (ECG) que complican su interpretación clínica. Asimismo, una alta impedancia aumenta la susceptibilidad a artefactos de movimiento, lo que puede distorsionar componentes críticos de la señal, como el complejo QRS en el ECG. Adicionalmente, esta condición puede limitar la capacidad de monitorear cambios fisiológicos relacionados con la piel, como los niveles de hidratación, afectando tanto mediciones basadas en sudor como estudios dermatológicos [5].<p>

<p>Finalmente, se ha reportado que las propiedades de la piel, más que las características del electrodo en sí, son las que ejercen mayor influencia sobre la impedancia de la interfaz y su estabilidad a lo largo del tiempo, lo que refuerza la importancia de esta etapa dentro del proceso de adquisición de biopotenciales [5].<p>

En relación con las estrategias para mejorar el acoplamiento piel-electrodo, uno de los enfoques más explorados ha sido la reducción de la impedancia de contacto. Para ello, algunos estudios han propuesto aumentar el área del electrodo, ya que, desde el punto de vista teórico, una mayor superficie de contacto permite disminuir la impedancia en la interfaz. Sin embargo, esta estrategia presenta limitaciones prácticas, especialmente en el contexto de dispositivos portátiles o miniaturizados, como relojes inteligentes, donde el tamaño del electrodo está restringido y no puede incrementarse indefinidamente [4].

Otra línea de investigación se ha centrado en el desarrollo y selección de nuevos materiales de electrodos, particularmente en el caso de electrodos secos, con el objetivo de mejorar su desempeño frente a los electrodos húmedos tradicionales. El enfoque clásico para evaluar estos materiales ha sido predominantemente empírico, basado en la realización de experimentos sobre piel humana y la comparación cualitativa o cuantitativa con electrodos húmedos, considerados el estándar de referencia. No obstante, este método presenta limitaciones importantes debido a la variabilidad inter e intrasujeto en las propiedades de la piel, lo que dificulta la reproducibilidad y generalización de los resultados. Como complemento, algunos trabajos han empleado modelos de impedancia equivalente de la interfaz electrodo-piel, evidenciando que el incremento del área del electrodo efectivamente reduce la impedancia, aunque sin resolver completamente las limitaciones prácticas asociadas [4].

Más recientemente, se han desarrollado soluciones avanzadas basadas en materiales compuestos y estructuras flexibles, orientadas a mejorar simultáneamente la conductividad eléctrica y la adaptabilidad mecánica del electrodo. Un ejemplo de ello es el uso de electrodos secos adaptables a la piel fabricados a partir de una película de polipirrol (PPy) integrada en un sustrato elastomérico de poliestireno-etileno-butadieno-estireno (SEBS), con nanocables de oro (Au NWs) alineados verticalmente. En este tipo de configuraciones, cada componente cumple una función específica: la película de PPy contribuye a reducir la impedancia de contacto (alcanzando valores del orden de 3 kΩ·cm² a 1 kHz) y a incrementar la capacidad de almacenamiento de carga; los nanocables de oro proporcionan baja resistencia superficial y alta elasticidad mecánica; mientras que el sustrato flexible favorece un contacto íntimo y confortable con la piel. Este tipo de enfoques integrales demuestra que la optimización del acoplamiento piel-electrodo no depende de un único factor, sino de la combinación de propiedades eléctricas y mecánicas que permitan una interfaz más estable y eficiente [6].

```python
import sys
import matplotlib.pyplot as plt
import numpy as np
import neurokit2 as nk
from ecgdetectors import Detectors
```

```python
# 1. Cargar datos
datos = np.loadtxt('P4ECG.txt')

# Grafica
plt.figure(figsize=(15, 5))
plt.plot(datos, linewidth=0.8) 
plt.title('Señal EMG - Vista Expandida')
plt.xlabel('Muestras')
plt.ylabel('Voltaje (V)')
plt.show()
```

```python
fs = 55  # Frecuencia de muestreo (ajusta este valor según tu sensor, ej: 100, 250, 500)

#Prueba de métodos
# MÉTODO 1: NeuroKit2 
# Limpia la señal y detecta picos
signals, info = nk.ecg_process(datos, sampling_rate=fs)
picos_nk = info['ECG_R_Peaks']

# MÉTODO 2: Py-ECG-Detectors
detectors = Detectors(fs)
picos_det = detectors.hamilton_detector(datos)


plt.figure(figsize=(15, 6))
plt.plot(datos, color='lightgray', label='Señal EKG', alpha=0.8)

# Marcar picos de NeuroKit
plt.scatter(picos_nk, datos[picos_nk], color='red', label='NeuroKit (Pan-Tompkins)', marker='o')

# Marcar picos de Detectors (con un pequeño desfase visual para comparar)
plt.scatter(picos_det, datos[picos_det], color='blue', label='Detectors (Hamilton)', marker='x')

plt.title('Comparación de Detectores de Picos R')
plt.legend()
plt.show()

print(f"Picos detectados por NeuroKit: {len(picos_nk)}")
print(f"Picos detectados por Detectors: {len(picos_det)}")
```

```python
# 2. Limpiar la señal de EKG (Elimina el vaivén de la línea base y ruido de 60Hz)
ecg_cleaned = nk.ecg_clean(datos, sampling_rate=fs, method="neurokit")

# 3. Detectar picos R usando el método de NeuroKit (el que te funcionó bien)
peaks, info = nk.ecg_peaks(ecg_cleaned, sampling_rate=fs, method="neurokit")

plt.rcParams['figure.figsize'] = [15, 10]
plt.rcParams['figure.dpi'] = 100 # Aumenta la resolución para que no se vea borroso 

# 4. Calcular HRV (Variabilidad de la Frecuencia Cardíaca)
hrv_indices = nk.hrv(peaks, sampling_rate=fs, show=True)


# 5. Mostrar los resultados principales (Corregido)
mean_nn = hrv_indices['HRV_MeanNN'].values[0]
bpm_medio = 60000 / mean_nn

print("--- Índices de HRV detectados ---")
print(f"Intervalo R-R promedio (MeanNN): {mean_nn:.2f} ms")
print(f"Frecuencia Cardíaca Media: {bpm_medio:.2f} BPM") # <--- ESTO es lo que el profe esperará ver
print(f"RMSSD (Actividad Parasimpática): {hrv_indices['HRV_RMSSD'].values[0]:.2f} ms")
print(f"SDNN (Variabilidad Total): {hrv_indices['HRV_SDNN'].values[0]:.2f} ms")

plt.tight_layout(pad=3.0) # pad aumenta el espacio entre los cuadros
plt.show()
```

```python
# Extraer índices reales de picos
peaks_indices = np.where(peaks["ECG_R_Peaks"] == 1)[0]

print("Número de picos reales:", len(peaks_indices))

if len(peaks_indices) < 2:
    print("No hay suficientes picos")
else:
    rr_intervals = np.diff(peaks_indices) * (1000 / fs)

    time_domain_features = get_time_domain_features(rr_intervals)

    print("--- Características del Dominio del Tiempo ---")
    for feature, value in time_domain_features.items():
        print(f"{feature:10}: {value:.2f}")
```
